# Feature Engineering & EDA
Feature Engineering adalah proses mengubah data mentah menjadi fitur (kolom) yang lebih
informatif dan sesuai untuk model machine learning, seringkali dengan memanfaatkan
pengetahuan domain untuk mengekstrak sinyal yang relevan. 

Sebagai ilustrasi, proses ini dapat
diibaratkan seperti seorang koki yang menyiapkan bahan mentah (memotong, membumbui)
sebelum memasak agar hasil akhirnya maksimal. Praktik ini sangat erat kaitannya dengan
Exploratory Data Analysis (EDA), yaitu proses investigasi awal pada dataset untuk
menemukan pola, mendeteksi anomali, dan merangkum karakteristik utama melalui statistik
dan visualisasi. 

Keduanya saling terkait erat: EDA memberikan wawasan untuk melakukan
feature engineering, dan fitur baru yang dibuat kemudian dianalisis kembali melalui EDA.

## 2. Encoding Data Kategorikal
Sebagian besar algoritma machine learning memerlukan input berupa data numerik. Oleh
karena itu, data kategorikal (teks) harus diubah menjadi representasi angka melalui proses yang
disebut encoding. Terdapat dua pendekatan utama:

#### 2.1. Label Encoding & Ordinal Encoding

Secara umum, Label Encoding adalah metode yang mengubah setiap kategori unik
menjadi sebuah nilai integer (misal: 'Merah' => 0, 'Biru' = >1). Namun, pendekatan ini
harus digunakan dengan hati-hati pada fitur, karena penetapan angka yang otomatis
(misalnya berdasarkan abjad) dapat menciptakan urutan yang tidak bermakna dan
menyesatkan model.

Untuk data yang memiliki tingkatan jelas (data ordinal), metode yang lebih tepat
adalah Ordinal Encoding. Metode ini memungkinkan kita untuk secara eksplisit
mendefinisikan urutan angkanya (contoh: 'Kecil' => 0, 'Sedang' => 1, 'Besar' => 2)
sehingga representasi numeriknya akurat dan sesuai dengan makna data.

#### Kolom sex memiliki dua kategori, sehingga cocok untuk Label Encoding. 
LabelEncoder mengubah 'Female' menjadi 0 dan 'Male' menjadi 1.

```python
le = LabelEncoder()
df['sex_encoded'] = le.fit_transform(df['sex'])
print(df[['sex', 'sex_encoded']].head())
```

#### 2.2. One-Hot Encoding

Metode ini mebuat kolom biner (bernilai 0 atau 1) baru untuk setiap kategori unik.
Pendekatan ini ideal untuk data nominal, yaitu data yang tidak memiliki urutan intrinsik
(misalnya: 'warna', 'kota'). Tujuannya adalah untuk menghindari asumsi adanya urutan
yang salah pada data, meskipun dapat menambah jumlah kolom secara signifikan jika
kategori sangat banyak.

#### Kolom island adalah fitur nominal. Terapkan One-Hot Encoding pada fitur ini.

```python
island_one_hot = pd.get_dummies(df['island'], prefix='island')
df = df.join(island_one_hot)
df.head()
```

```python
avg_mass = df.groupby('species')['body_mass_g'].mean().sort_values()

print("Urutan species berdasarkan massa rata-rata:")
print(avg_mass)

species_order = avg_mass.index.tolist()

ordinal_encoder = OrdinalEncoder(categories=[species_order])

df['species_ordinal'] = ordinal_encoder.fit_transform(df[['species']])

print(df[['species', 'species_ordinal']].sample(10, random_state=1))
```

## 3. Binning (Discretization)
Binning, atau dikenal juga sebagai discretization, adalah sebuah teknik preprocessing data
untuk mentransformasi fitur numerik kontinu menjadi fitur kategorikal diskrit. 

Proses ini dilakukan dengan membagi rentang nilai dari sebuah fitur ke dalam sejumlah interval atau
"bins", kemudian menetapkan setiap titik data ke dalam bin yang sesuai. Sebagai contoh, fitur
'harga' dapat dikelompokkan ke dalam kategori 'Rendah', 'Menengah', dan 'Tinggi'.

Tujuan dari proses ini adalah untuk menyederhanakan data, meningkatkan robustisitas
model terhadap nilai ekstrem (outlier), dan memfasilitasi algoritma machine learning dalam
menangkap hubungan non-linear antar variabel.

```python
df.describe()
```

```python
bins = [2500, 4000, 5000, 6500]
labels = ['Ringan', 'Sedang', 'Berat']

df['weight_category'] = pd.cut(df['body_mass_g'], bins=bins, labels=labels)

print(df[['body_mass_g', 'weight_category']].sample(10, random_state=1))
```

```python
df['weight_category'] = pd.cut(df['body_mass_g'], bins=bins,
labels=labels)
print(df[['body_mass_g', 'weight_category']].sample(5))
print("\nDistribusi Kategori Berat:")
print(df['weight_category'].value_counts())
print("\nInformasi DataFrame setelah binning:")
df.info()
```

## 4. Feature Selection
Feature Selection adalah proses seleksi sistematis untuk memilih subset fitur (variabel atau
kolom) yang paling relevan dari sebuah dataset. 

Tujuan utamanya bukan sekadar mengurangi
jumlah dimensi data, melainkan untuk meningkatkan performa prediksi model, mengurangi
risiko overfitting (ketika model terlalu hafal data training), mempercepat waktu komputasi,
serta meningkatkan interpretabilitas model.

Secara umum, metode feature selection dapat diklasifikasikan ke dalam tiga kategori utama:
- **Metode Filter** mengevaluasi relevansi fitur berdasarkan karakteristik statistik data, terlepas dari
model machine learning yang akan digunakan.
- **Metode Wrapper** menggunakan performa sebuah model machine learning spesifik untuk
mengevaluasi dan memilih subset fitur terbaik.
- **Metode Embedded** mengintegrasikan proses seleksi fitur langsung ke dalam algoritma saat
model sedang dilatih (training).

```python
correlation_matrix = df.corr(numeric_only=True)
plt.figure(figsize=(12, 10))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm',
fmt=".2f")
plt.title('Heatmap Korelasi Fitur Numerik')
plt.show()
```

## 5. Exploratory Data Analysis (EDA) dan Visualisasi Distribusi
Exploratory Data Analysis (EDA) adalah pendekatan investigasi untuk menganalisis dan
merangkum sebuah dataset. 

Tujuan utamanya adalah untuk menemukan pola, mendeteksi
anomali, menguji hipotesis, dan mengidentifikasi karakteristik utama data, seringkali dengan
bantuan statistik deskriptif dan representasi grafis.

```python
sns.histplot(data=df, x='body_mass_g', hue='species', multiple='stack', bins=30)
plt.title('Distribusi Body Mass G berdasarkan Species')
plt.show()
```

```python
sns.boxplot(data=df, x='species', y='body_mass_g')
plt.title('Boxplot Body Mass G berdasarkan Species')
plt.show()
```

```python
sns.catplot(data=df, x='species', y='body_mass_g', kind='box',
col='island', hue='sex')
plt.show()
```
