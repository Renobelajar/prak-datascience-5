## 3. Data Cleaning & Preprocessing

### 3.1 Data Cleaning

```python
print(df.isnull().sum())
```

```python
df["age"] = df["age"].fillna(df["age"].median())
```

```python
print(df.isnull().sum())
```

```python
df.drop("deck", axis=1, inplace=True)
```

```python
print(df.isnull().sum())
```

```python
df["embarked"] = df["embarked"].fillna(df["embarked"].mode()[0])
df["embark_town"] = df["embark_town"].fillna(df["embark_town"].mode()[0])
```

```python
print(df.isnull().sum())
```

```python
print(df.duplicated().sum())
```

```python
df.drop_duplicates(inplace=True)
```

```python
print(df.duplicated().sum())
```

### 3.2 Data Preprocessing

#### Normalisasi dengan MinMaxScaler

```python
scaler_minmax = MinMaxScaler()

df["fare_normalized"] = scaler_minmax.fit_transform(df[["fare"]])
print(df[["fare", "fare_normalized"]].head())
```

#### Standarisasi dengan StandardScaler

```python
scaler_standard = StandardScaler()

df["fare_standarized"] = scaler_standard.fit_transform(df[["fare"]])
print(df[["fare", "fare_standarized"]].head())
```

```python
df.to_csv("titanic_clean.csv", index=False)
```

# Scraping Data CNN

```python
import pandas as pd
import time
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
```

```python
# Fungsi Mencari URL Pencarian

def get_url(page):
    if page == 1:
        return "https://www.cnnindonesia.com/search?query=IHSG"
    else:
        return f"https://www.cnnindonesia.com/search?query=IHSG&page={page}"
```

```python
# Fungsi menyimppan data ke CSV
def save_to_csv(titles):
 
    df = pd.DataFrame(titles, columns=['judul'])
    df.to_csv('judul_berita_ihsg.csv', index=False,encoding='utf-8-sig')
    print(f'Sukses simpan {len(df)} judul berita ')
```

```python
# Fungsi Scraping Judul Berita

def scrape_titles():
    options = Options()
    options.add_argument('--disable-gpu')
    driver = webdriver.Chrome(options=options)
    titles = []

    for page in range(1, 11):
        url = get_url(page)
        print(f'Scraping {url}')

        driver.get(url)
        time.sleep(2) # Tunggu halaman selesai render

        try:
            h2_elements = driver.find_elements(By.CLASS_NAME,'text-cnn_black_light')
            if not h2_elements:
                print(f'Tidak ada judul ditemukan di halaman{page}.')
                break
            for h2 in h2_elements:
                titles.append(h2.text)
        except Exception as e:
                print(f'Error di halaman {page}: {e}')
                continue
    driver.quit()
    return titles


hasil_judul = scrape_titles()
save_to_csv(hasil_judul)
```
